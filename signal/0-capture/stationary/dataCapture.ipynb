# Stationary Device: Data Capture

Initially, a six-face (i.e. gravity vector parallel/antiparallel to one axis at a time) approach was used, but due to imperfect measurements this has been found to produce small errors in the affine transformation used for acceleration calibration, which in turn produces further problems down the pipeline. This can be mitigated by capturing with the device tilted by a known angle in a variety of orientations (gravity vector decomposed in two axes at a time, yielding 8 combinations for a given angle) to overdetermine the calibration parameters. 

Materials: 
* Lectern or music stand
* 45/45 or 30/60 drafting triangles or protractor
* Flat horizontal and vertical surfaces
* Device

<img src="./orientation-reference/materials.jpg" alt="Materials" width="400" height="400">

How to run the test:

0. Make sure the device isn't performing any form of calibration. Flash it to NOT do so if necessary.
1. Turn on the device and let it heat up a little bit.
2. Select the Python kernel from `signal/venv`.
3. Run the setup helper code and connect to the device.
4. Choose a common tilt angle (recommended: 45 degrees with respect to the horizontal) and setup the lectern or music stand.
5. For each orientation section: 

    5.1. Trigger notification streaming on the device.

    5.2. Place it as it's shown in the reference image. Use the HARDWARE axes, as drawn in the sensor. 
    
    5.3. Trigger the capture cell.

6. Disconnect from the device and turn it off.

## Setup

```python
from pathlib import Path
from signal_utils import IMUSampleReceiver

LISTEN_DURATION_SECONDS = 10.0
G = 9.80665

OUTPUT_DIR = Path("./jul-25-2026").resolve()

receiver = IMUSampleReceiver(listenDurationSeconds=LISTEN_DURATION_SECONDS)


async def sampleOrientation(fileName: str):
    if not receiver.isConnected():
        raise RuntimeError("Run the connect cell before sampling.")

    filePath = OUTPUT_DIR / fileName
    print(f"Saving capture to {filePath}")
    return await receiver.receive(exportedFileName=str(filePath))


print(f"CSV output directory: {OUTPUT_DIR}")
```

## Connect

```python
if receiver.isConnected():
    print("Receiver already connected")
else:
    await receiver.connect()
```
