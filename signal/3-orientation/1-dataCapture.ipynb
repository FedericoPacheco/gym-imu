# Complementary filter: Data capture

Use this notebook to collect simple rotations to understand filtering results.

How to run the test: 

1. Turn on the device and let it heat up a little bit.
2. Select the Python kernel from `signal/venv`.
3. Run the setup helper code and connect to the device.
4. For each axis, perform the following rotations around it in a smooth and relatively slow manner:
    * +45 degrees, pause, go back, pause.
    * +90 degrees, pause, go back, pause.
    * +180 degrees, pause, go back, pause.
    * +360 degrees, pause, go back, pause.

Notes: 
* **Make sure the device is positioned in its normal operating orientation, i.e. being worn on the chest. Do NOT orient the captured axis in the same line of gravity**, as it was done on the six-face test for calibration. This makes filter validation harder. 
* **Pause for about 5 seconds between each movement** to make the interpretation of the results easier. 
* The **positive direction of rotation** is defined as the one that follows the **right-hand rule**. The **system of coordinates** for xyz axes is the one engraved in the 1st prototype case, **RDF (right, down, forwards)**. 

## Setup

```python
from pathlib import Path
from signal_utils import IMUSampleReceiver

LISTEN_DURATION_SECONDS = 120

OUTPUT_DIR = Path("./capture").resolve()

receiver = IMUSampleReceiver(listenDurationSeconds=LISTEN_DURATION_SECONDS)


async def sampleRotation(fileName: str):
    if not receiver.isConnected():
        raise RuntimeError("Run the connect cell before sampling.")

    filePath = OUTPUT_DIR / fileName
    print(f"Saving capture to {filePath}")
    return await receiver.receive(exportedFileName=str(filePath))


print(f"CSV output directory: {OUTPUT_DIR}")
```

## Connect

```python
if receiver.isConnected():
    print("Receiver already connected")
else:
    await receiver.connect()
```

## Around z axis (XY plane)

```python
rotationZData = await sampleRotation("rotation-z-hw.csv")
```

## Around y axis (XZ plane)

```python
rotationYData = await sampleRotation("rotation-y-hw.csv")
```

## Around x axis (YZ plane)

```python
rotationXData = await sampleRotation("rotation-x-hw.csv")
```

## Disconnect

```python
await receiver.disconnect()
```
