# Complementary filter: Data capture

Use this notebook to collect simple rotations to understand filtering results.

How to run the test: 

Follow almost the same steps as in the calibration data capture notebook. Then perform the following smooth and relatively slow rotations around each axis:
* +45 degrees, pause, go back, pause
* +90 degrees, pause, go back, pause
* +180 degrees, pause, go back, pause
* +360 degrees, pause, go back, pause

Place the device on a table with relatively low friction. Pause for about 5 seconds between each movement to make the interpretation of the results easier. The positive direction of the rotation is defined as the one that follows the right-hand rule. The system of coordinates for xyz is the one engraved in the 1st prototype case, RDF (right, down, forwards). 

## Setup

```python
from pathlib import Path
from signal_utils import IMUSampleReceiver

LISTEN_DURATION_SECONDS = 120

OUTPUT_DIR = Path("./capture").resolve()

receiver = IMUSampleReceiver(listenDurationSeconds=LISTEN_DURATION_SECONDS)


async def sampleRotation(fileName: str):
    if not receiver.isConnected():
        raise RuntimeError("Run the connect cell before sampling.")

    filePath = OUTPUT_DIR / fileName
    print(f"Saving capture to {filePath}")
    return await receiver.receive(exportedFileName=str(filePath))


print(f"CSV output directory: {OUTPUT_DIR}")
```

## Connect

```python
if receiver.isConnected():
    print("Receiver already connected")
else:
    await receiver.connect()
```

## Around z axis (XY plane)

```python
rotationZData = await sampleRotation("rotation-z.csv")
```

## Around y axis (XZ plane)

```python
rotationYData = await sampleRotation("rotation-y.csv")
```

## Around x axis (YZ plane)

```python
rotationXData = await sampleRotation("rotation-x.csv")
```

## Disconnect

```python
await receiver.disconnect()
```
