# Ideas on how to implement repetition detection

If repetitions can be identified reliably, they may help define brief integration windows for velocity, which paired with ZUPTs could help stabilize estimates.  

Possible methods/tools:

* High-pass filter: attempt to get only spikes or rapid changes in acceleration, filtering out low-frequency remaining error/noise. Cons: may not make sense, since genuine low acceleration intervals are not necessarily stationary intervals. Additionally, it will distort the signal, especially if the filter has nonlinear phase (like a cheap, streaming friendly, recursive filter), or just with gibbs-like oscillations.

* Peak detection: find local maxima similarly to *scipy.signal.find_peaks()*. Pros: simple. Cons: not sure it will work at all, because each repetition has multiple peaks, at least with norms.

* State machine: define states like *rest*, *concentric*, *isometric/end*, *eccentric* and transition between them based on some underlying pattern in each exercise's acceleration, angular velocity, or orientation. Require thresholds + minimum durations. Pros: real-time friendly, but will require buffering. Cons: decision criteria/parameters for states.

* Template matching: define a standard repetition kernel from a representative repetition and correlate it with the signal to detect complete repetitions. As a quick analysis suggests, this kernel changes with different exercises, and maybe even if the technique is modified. Pros: accurate. Cons: exercise-specific, sensitive, not that streaming/real-time friendly.

* Machine learning: train a model to recognize repetitions based on features extracted from the sensor data. Pros: most accurate, probably. Cons: computationally intensive.

In all cases manual labeling of repetitions is probably required. 

Making a general repetition detection algorithm will likely be challenging.

Multiple methods may be assembled together.

Implementation will be deferred for now.
